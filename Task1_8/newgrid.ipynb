```python
import numpy as np
import h5py


def gauss_seidel(f):
    newf = f.copy()
    for i in range(1, newf.shape[0] - 1):
        for j in range(1, newf.shape[1] - 1):
            newf[i, j] = 0.25 * (newf[i, j+1] + newf[i, j-1] +
                                 newf[i+1, j] + newf[i-1, j])
    return newf


def initialize(grid_sizes):
    f_numpy = np.random.rand(grid_sizes, grid_sizes).astype(np.float64)
    f_numpy[0, :] = f_numpy[-1, :] = f_numpy[:, 0] = f_numpy[:, -1] = 0
    return f_numpy


def save_grid(grid, filename):
    with h5py.File(filename, 'w') as f:
        f.create_dataset('New Grid', data=grid)

def read_grid(filename):
    with h5py.File(filename, 'r') as f:
        return f['New Grid'][:]


    

if __name__ == "__main__":
    grid_size = 5
    f_numpy = initialize(grid_size)
    newgrid = gauss_seidel(f_numpy)
    save_grid(newgrid, 'newgrid.hdf5')
    read_grid = read_grid('newgrid.hdf5')
    print("New Grid (Computed):")
    print(newgrid)
    print("\nRead from File:")
    print(read_grid)
    


    

```
